The Superstore dataset was successfully loaded and inspected. The dataset contains 9,994 rows and 21 columns.
```python
import pandas as pd
```

```python
df = pd.read_csv("Superstore_EDA.csv" , encoding=("latin1"))
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
```

```python
df["Order Date"].dtype
```

```python
df["Month"] = df["Order Date"].dt.to_period("M")
df["Quarter"] = df["Order Date"].dt.to_period("Q")
```

```python
df[["Order Date", "Month", "Quarter"]].head()
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()
```

```python
monthly_sales
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

### Observation

The monthly sales trend shows fluctuations in sales across different months. Some months experience noticeably higher sales than others, indicating seasonal variations in customer purchasing behaviour. The trend will be examined further using quarterly sales analysis to identify broader patterns.

```python
quarterly_sales = df.groupby("Quarter")["Sales"].sum()
```

```python
quarterly_sales
```

```python
plt.figure(figsize=(10,5))

plt.plot(quarterly_sales.index.astype(str), quarterly_sales.values ,marker="o")

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()
```

### Observation

The quarterly sales trend shows variations in sales performance across different quarters. Overall, sales demonstrate an upward pattern over the years, although individual quarters show fluctuations. This indicates both long-term growth and seasonal variation in sales.

```python
segment_sales = df.groupby("Segment")["Sales"].sum()
```

```python
segment_sales
```

```python
plt.figure(figsize=(8,5))

plt.bar(segment_sales.index, segment_sales.values)

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

### Observation

The Consumer segment generates the highest total sales among the three customer segments, followed by Corporate and Home Office. This indicates that Consumer customers contribute the largest share of sales and may represent an important customer group for the business.

### Customer Demographics Limitation

The dataset does not contain customer age or gender information. Therefore, age-group distribution and gender-based analysis cannot be performed reliably. Instead, customer segments were analyzed using the available Segment column.

```python
product_sales=df.groupby("Product Name")["Sales"].sum()

top_10_products = product_sales.sort_values(ascending=False).head(10)
```

```python
plt.figure(figsize=(10, 6))

plt.barh(top_10_products.index[::-1], top_10_products.values[::-1])

plt.title("Top 10 Best-Selling Products")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")

plt.tight_layout()
plt.show()
```

### Observation

The top 10 products contribute a significant amount of sales to the Superstore dataset. The chart shows that sales are concentrated among a relatively small group of products, which can help identify products that may deserve greater attention in inventory planning and sales strategies.

```python
category_sales = df.groupby("Category")["Sales"].sum()
```

```python
category_sales
```

```python
plt.figure(figsize=(8,5))

plt.bar(category_sales.index, category_sales.values)

plt.title("Revenue by product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

### Observation

Technology generates the highest total sales among the three product categories, followed by Furniture and Office Supplies. This indicates that Technology products contribute a substantial portion of the overall revenue.

```python
numeric_data = df[["Sales","Quantity","Discount","Profit"]]

correlation_matrix = numeric_data.corr()
```

```python
correlation_matrix
```

```python
import seaborn as  sns
```

```python
plt.figure(figsize=(18,6))

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()

plt.show
```

### Observation

The correlation heatmap shows the relationships between Sales, Quantity, Discount, and Profit. Sales and Profit show a positive relationship, indicating that higher sales are generally associated with higher profit. The relationship between Discount and Profit should also be examined carefully, as higher discounts may be associated with changes in profitability.

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="Discount", y="Profit")

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.tight_layout()
plt.show()
```

### Observation

The scatter plot shows the relationship between discount and profit. Higher discount levels are associated with greater variation in profit, including several negative-profit transactions. This suggests that excessive discounting may negatively affect profitability for some sales.

## Business Recommendations

Based on the analysis, the following recommendations can be made:

1. **Focus on high-performing products:** The business should maintain sufficient inventory and promote products that consistently generate high sales, particularly those appearing among the top-selling products.

2. **Strengthen the Technology category:** Technology generates the highest sales among the product categories, so the business can consider expanding its product range and promotional activities in this category.

3. **Monitor discount strategies:** The analysis shows variation in profitability at different discount levels. The business should carefully evaluate discount levels to avoid reducing profit margins unnecessarily.

4. **Use customer segmentation:** The Consumer segment contributes the largest sales, so targeted promotions and retention strategies can be developed based on the purchasing behaviour of different customer segments.

## Conclusion

The exploratory data analysis of the Superstore dataset provided insights into sales trends, customer segments, and product performance. Monthly and quarterly analysis revealed variations in sales over time, while the customer segment analysis showed that the Consumer segment contributes the highest sales. Product analysis identified the top-performing products and showed that Technology generates the highest sales among the three categories. The correlation and discount analysis also provided insights into the relationships between sales, profit, quantity, and discount.

Overall, the analysis demonstrates how exploratory data analysis can be used to understand business performance and support data-driven decision-making.
